```python
df = pd.read_csv("club_membership.csv")

df.isnull().sum()
```

```python
total_students = len(df)

print("Total Students:", total_students)
```

```python
club_names = df['Club'].unique()

print(club_names)
```

```python
club_counts = df['Club'].value_counts()

print(club_counts)
```

```python
membership = np.array(club_counts.values)

average_members = np.mean(membership)

print("Average Club Membership:", average_members)
```

```python
club_members = df.groupby('Club')['Student_ID'].count()

print(club_members)
```

```python
most_popular = club_members.idxmax()

print("Most Popular Club:", most_popular)
```

```python
least_popular = club_members.idxmin()

print("Least Popular Club:", least_popular)
```

```python
percentage = (club_members / len(df)) * 100

print(percentage)
```

```python
club_analysis = pd.DataFrame({
    'Total_Members': club_members,
    'Percentage': percentage
})

print(club_analysis)
```

```python
club_analysis.to_csv("club_analysis.csv")
```

```python
club_analysis.to_csv("club_analysis.csv", index=True)

print("Analysis file saved successfully!")
```

```python
plt.figure(figsize=(8,5))

plt.bar(club_analysis.index,
        club_analysis['Total_Members'])

plt.title("College Club Membership Breakdown")
plt.xlabel("College Clubs")
plt.ylabel("Number of Students")

plt.xticks(rotation=30)
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x='Club')

plt.title("Number of Students in Each Club")
plt.xlabel("College Clubs")
plt.ylabel("Number of Students")

plt.xticks(rotation=30)
plt.show()
```

```python
club_year = pd.crosstab(df['Club'], df['Year'])

plt.figure(figsize=(8,5))

sns.heatmap(club_year, annot=True, cmap='Blues', fmt='d')

plt.title("Club Membership by Academic Year")
plt.show()
```
